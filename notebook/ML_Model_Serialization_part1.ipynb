###IMPORTS

```python
import pickle
import joblib
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

## Read the Data

```python
df=pd.read_csv('./content/housing.csv')
```

```python
df
```

```python
updated_df=df.drop('ocean_proximity',axis=1)
```

```python
updated_df
```

```python
updated_df=updated_df.dropna()
```

```python
updated_df
```

```python
X=updated_df.drop(columns='median_house_value')
y=updated_df['median_house_value']
```

```python
X
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_predict=model.predict(X_test)
r2score= r2_score(y_test, y_predict)
r2score
```

## PICKLE Integration

```python
with open('model.pkl','wb') as f:
  pickle.dump(model,f)
```

```python
with open('model.pkl','rb') as f:
  pickle_model=pickle.load(f)
```

```python
pickle_model
```

```python
pickle_predit=pickle_model.predict(X_test)
```

```python
r2=r2_score(y_test,pickle_predit)
r2
```

###Joblib

```python
joblib.dump(model,'modelJL.joblib')
```

```python
joblib_model=joblib.load('modelJL.joblib')
```

```python
joblib_predit=joblib_model.predict(X_test)
```

```python
r2=r2_score(y_test,joblib_predit)
r2
```
